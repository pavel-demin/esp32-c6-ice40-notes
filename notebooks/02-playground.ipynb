### Playground

This notebook shows how to communicate with configuration registers, status registers and AXI4-Stream interfaces inside the FPGA using the [pyhubio](https://github.com/pavel-demin/pyhubio) library.

Import the required libraries and instantiate `PyhubESP`.

```python
from pyhubio import PyhubESP
import numpy as np
import time

io = PyhubESP("192.168.4.1")
```

Start I/O and program the FPGA.

```python
io.start()
io.program("playground.bin")
```

The port number (hub address) of the configuration registers is 0.

The first bit of the 32-bit configuration register at address 0 is connected to the LED on the development board.

The following script turns the LED on and off several times.

```python
leds = np.zeros(1, np.uint32)

for i in range(10):
    leds[0] ^= 1
    io.write(leds, port=0, addr=0)
    time.sleep(0.5)
```

The 32-bit configuration register at address 1 is connected to the 16-bit inputs of the DSP multiplier and the status register at address 1 is connected to the 32-bit output of the multiplier.

The following script writes two 16-bit integer values to be multiplied in the configuration register at address 1 and reads the result of the multiplication from the status register at address 0.

```python
input = np.array([6, 7], np.int16)
io.write(input, port=0, addr=1)

output = np.zeros(1, np.int32)
io.read(output, port=1, addr=0)

print(output[0])
```

The first AXI4-Stream interface has port number (hub address) 2. Its slave side is connected to a counter that increments after each read transaction.

The following script reads 10 numbers from the counter.

```python
output = np.zeros(10, np.int32)
io.read(output, port=2, addr=0)

print(output)
```

Stop I/O.

```python
io.stop()
```
