### LED blinker

This first notebook shows how to program the FPGA on the development board using the [pyhubio](https://github.com/pavel-demin/pyhubio) library.

Import the `PyhubESP` class from the [pyhubio](https://github.com/pavel-demin/pyhubio) library and instantiate it.

```python
from pyhubio import PyhubESP

io = PyhubESP("192.168.4.1")
```

Program the FPGA using the `led_blinker.bin` file.

```python
io.start()
io.program("led_blinker.bin")
io.stop()
```

If the FPGA is programmed successfully, the LED on the development board should start blinking.
